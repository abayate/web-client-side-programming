```python
%%writefile nasa.py
# apod.py
import requests
from IPython.display import display, Image

def get_apod(api_key):
    response = requests.get(f'https://api.nasa.gov/planetary/apod?api_key={api_key}')
    if response.status_code == 200:
        data = response.json()
        return data['title'], data['url'], data['explanation'], data.get('media_type', 'image')
    else:
        return "Could not fetch APOD data", "", "", ""

def display_apod(api_key):
    title, url, explanation, media_type = get_apod(api_key)
    if title == "Could not fetch APOD data":
        print(title)
    else:
        print(f"Title: {title}\nURL: {url}\nExplanation: {explanation}")
        # fixed: some days the APOD is a video, which Image() cannot show
        if media_type == "image":
            display(Image(url=url, embed=True))
        else:
            print(f"(Today's APOD is a {media_type}, open the URL above to watch it.)")
```

## **Find any other OpenAI API feature and demonstrate its working**

**Feature: Embeddings (semantic search).** An embedding turns text into a vector of numbers; texts with similar meaning get vectors that point the same way. Below I embed a few of my course topics and rank them by cosine similarity to a question, which is how RAG apps find relevant documents.

```python
import math

docs = [
    "Vue uses reactive refs and computed properties to update the page.",
    "Svelte compiles components to plain JavaScript at build time.",
    "The NASA APOD API returns a new astronomy picture every day.",
    "Python functions can return multiple values as a tuple.",
    "A golden retriever is a friendly family dog.",
]
question = "Which JavaScript framework does its work at compile time?"

resp = client.embeddings.create(model="text-embedding-3-small", input=docs + [question])
vectors = [d.embedding for d in resp.data]
q_vec, doc_vecs = vectors[-1], vectors[:-1]

def cosine(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    return dot / (math.sqrt(sum(x * x for x in a)) * math.sqrt(sum(y * y for y in b)))

print(f"Anthony Bayate - embeddings demo ({len(q_vec)} dimensions per vector)")
print(f"Question: {question}\n")
for score, doc in sorted(((cosine(q_vec, v), d) for v, d in zip(doc_vecs, docs)), reverse=True):
    print(f"{score:.3f}  {doc}")
```
